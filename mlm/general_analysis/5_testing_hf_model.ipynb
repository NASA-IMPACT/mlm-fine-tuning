```python
import os
from transformers import AutoModel, AutoTokenizer, pipeline
from dotenv import load_dotenv
load_dotenv()
from huggingface_hub import login
login(os.getenv("HUGGINGFACE_TOKEN"))
```

```python

# Load model or dataset
model_name = "nasa-impact/indus-sde-v0.1"
# tokenizer = AutoTokenizer.from_pretrained(model_name, )
# model = AutoModel.from_pretrained(model_name)



# Load the fill-mask pipeline
fill_mask = pipeline("fill-mask", model=model_name)

# Example input for the fill-mask task
text = "The quick brown fox jumps over the <mask> dog."

# Perform the fill-mask task
results = fill_mask(text)

# Display the results
for result in results:
    print(f"Token: {result['token_str']}, Score: {result['score']:.4f}")

```
