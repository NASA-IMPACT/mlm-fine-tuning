```python
import pandas as pd
```

```python
ej_text = pd.read_parquet('../data/ej_train_data.parquet')[["text"]]
non_ej_text = pd.read_parquet('../data/parsed_sde_init_check_with_title.parquet')[["text"]].sample(ej_text.shape[0])

mixed_text = pd.concat([ej_text, non_ej_text]).sample(frac=1, random_state=42)

mixed_text.to_parquet('../data/mixed_ej_data.parquet')
```
