```python
from transformers import AutoTokenizer, DataCollatorForLanguageModeling
from datasets import Dataset

# Mock Dataset for Demonstration
data = {
    "text": [
        "Transformers are amazing for NLP tasks.",
        "Masked language modeling is a great pretraining task.",
        "Tokenizers handle input preprocessing efficiently."
    ]
}
dataset = Dataset.from_dict(data)

# Load Tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Tokenize Dataset
def tokenize_function(examples):
    return tokenizer(examples["text"], truncation=True, padding="max_length", max_length=32)

tokenized_dataset = dataset.map(tokenize_function, batched=True)

# Initialize Data Collator
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=True,
    mlm_probability=0.15,  # Probability of masking tokens
)

# Create a Batch
batch = data_collator([tokenized_dataset[i] for i in range(3)])  # Use 3 examples

# Decode and Display Results
print("Masked Input IDs:", batch["input_ids"])
print("Labels:", batch["labels"])

# Decode Inputs and Labels for Human Readability
decoded_inputs = [tokenizer.decode(ids, skip_special_tokens=True) for ids in batch["input_ids"]]
decoded_labels = [tokenizer.decode(ids, skip_special_tokens=True) for ids in batch["labels"]]

print("\nDecoded Inputs with Masked Tokens:")
for text in decoded_inputs:
    print(text)

print("\nDecoded Labels (Ground Truth):")
for label in decoded_labels:
    print(label)
```
